```sql
SELECT
    TRIM(_c0) AS brand,
    TRIM(_c1) AS device,
    TRIM(_c2) AS model,
    TRIM(_c3) AS country_of_origin,
    TO_DATE(TRIM(_c4), 'dd-MM-yyyy') AS release_date,
    TRY_CAST(TRIM(_c5) AS DOUBLE) AS price_usd
FROM workspace.default.bronze_products
WHERE _c0 IS NOT NULL
  AND LOWER(TRIM(_c0)) NOT IN
  ('excel sample data', 'technological products data', 'brand')
  AND _c1 IS NOT NULL;
```

```sql
CREATE OR REPLACE TABLE workspace.default.silver_products AS

SELECT
    TRIM(_c0) AS brand,
    TRIM(_c1) AS device,
    TRIM(_c2) AS model,
    TRIM(_c3) AS country_of_origin,
    TRIM(_c4) AS release_date,
    TRY_CAST(TRIM(_c5) AS DOUBLE) AS price_usd

FROM workspace.default.bronze_products

WHERE _c0 IS NOT NULL
AND LOWER(TRIM(_c0)) NOT IN (
    'excel sample data',
    'technological products data',
    'brand'
)
AND _c1 IS NOT NULL;
```

```sql
CREATE OR REPLACE TABLE workspace.default.gold_brand_analysis AS
SELECT
    brand,
    COUNT(*) AS total_products,
    ROUND(AVG(CAST(price_usd AS DOUBLE)), 2) AS avg_price,
    MIN(CAST(price_usd AS DOUBLE)) AS minimum_price,
    MAX(CAST(price_usd AS DOUBLE)) AS maximum_price
FROM workspace.default.silver_products
GROUP BY brand;
```

```sql
SELECT *
FROM workspace.default.gold_brand_analysis;
```

```sql
SELECT * 
FROM workspace.default.gold_brand_analysis ORDER BY avg_price DESC;
```

```sql
SELECT
    brand,
    total_products
FROM workspace.default.gold_brand_analysis
ORDER BY total_products DESC;
```

```sql
SELECT MIN(avg_price) AS min_avg_price,
       MAX(avg_price) AS max_avg_price
FROM workspace.default.gold_brand_analysis;
```

```sql
WITH ranked AS (
    SELECT
        brand,
        avg_price,
        ROW_NUMBER() OVER (ORDER BY avg_price ASC) AS min_rank,
        ROW_NUMBER() OVER (ORDER BY avg_price DESC) AS max_rank
    FROM workspace.default.gold_brand_analysis
)
SELECT
    brand,
    avg_price,
    CASE
        WHEN min_rank = 1 THEN 'Minimum'
        WHEN max_rank = 1 THEN 'Maximum'
    END AS price_type
FROM ranked
WHERE min_rank = 1 OR max_rank = 1;
```
